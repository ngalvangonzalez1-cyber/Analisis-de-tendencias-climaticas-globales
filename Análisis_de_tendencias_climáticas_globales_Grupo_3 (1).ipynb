## **Proceso completo de limpieza y preparación de los datos**
Para desarrollar un análisis enfocado en el problema planteado inicialmente, se seleccionaron estratégicamente 21 variables del dataset original y se ajustó el período de estudio al rango 2004-2024. Al examinar en detalle las variables dentro de este rango, no se identificó ningún tipo de inconsistencia: los tipos de dato, las unidades de medida y los nombres de las columnas se encontraban correctamente definidos.

Con el fin de desarrollar el ejercicio de limpieza solicitado en clase, se introdujeron artificialmente errores controlados en el dataset, utilizando funciones de generación aleatoria (`numpy.random`). Tras esta modificación, se realizó un nuevo diagnóstico del dataset, se identificaron y solucionaron los siguientes problemas:

*   **Duplicados:** Se eliminaron 22 filas duplicadas mediante `drop_duplicates()`.
*   **Años fuera de rango:** Al revisar los valores máximo y mínimo de la columna `year`, se encontraron 38 registros con años imposibles, como 2099. Se eliminó cualquier fila cuyo año estuviera fuera del rango establecido (2004-2024).
*   **Nombres de país inconsistentes:** Existían 102 formas distintas de escribir el nombre de un país, cuando en realidad solo hay 30 países en el dataset. Esta inconsistencia se debía a espacios al inicio o al final del texto y a diferencias de mayúsculas y minúsculas. Se corrigió eliminando los espacios sobrantes con `.str.strip()` y colocando solo la primera letra de cada palabra **en** mayúscula, mediante `.str.title()`.
*   **Tipos de dato mezclados en `primary_energy_twh`:** Los valores de esta columna combinaban una parte numérica con la unidad de medida en texto (por ejemplo, "314.0 TWh"), lo que impedía realizar cálculos matemáticos. Se eliminó la parte de texto con `.str.replace(' TWh', '', regex=False)` y se convirtió la columna a tipo numérico, dado que la unidad de medida ya queda indicada en el nombre de la variable.
*   **Negativos inválidos en `share_renewables_pct`:** Al tratarse de un valor porcentual respecto a la implementación de energías renovables, no tenía sentido que existieran valores negativos; se detectaron 76 casos. Se corrigieron aplicando valor absoluto con `.abs()`.
*   **Negativos en `methane_emissions` y `nitrous_oxide`:** Estos valores negativos no se modificaron. Tras revisar la documentación del dataset, se determinó que estas cifras pueden representar procesos de captura o absorción neta, por lo que no representan un error.
*   **Valores atípicos en `temp_anomaly_global`:** De acuerdo con la información proporcionada por el autor del dataset, esta variable debía tomar valores entre 0 y 3, pero se encontraron registros por fuera de ese rango. Estos valores se marcaron como faltantes (`NaN`) para tratarlos junto con el resto de los datos nulos.
*   **Imputación de valores nulos:** se detectaron 507 valores faltantes entre las columnas `co2_emissions`, `share_renewables_pct`, `temp_anomaly_land` y `temp_anomaly_global`. Estos vacíos se completaron utilizando la mediana correspondiente a cada país y año, en lugar de un promedio general, dado que varias de estas variables presentan un comportamiento anual. Las columnas `sea_surface_temp_anom` y `ocean_heat_content` conservan valores nulos, pero estos corresponden a países sin costa.

Como resultado de estos siete pasos de limpieza, el dataset pasó de 7582 a 7522 filas, lo que representa una reducción inferior al 1%. Este bajo porcentaje de eliminación es un buen indicador de la calidad del proceso: se corrigieron los errores sin sacrificar información relevante.

````python
#ENSUCIAR
np.random.seed(42)
df_sucio = df_limpio.copy()
df_sucio = df_sucio.reset_index(drop=True)
n = len(df_sucio)

idx_todos = np.random.choice(df_sucio.index, size=int(n * 0.12), replace=False)

idx_duplicados     = idx_todos[0             : int(n*0.003)]
idx_nulos_co2      = idx_todos[int(n*0.003)  : int(n*0.033)]
idx_nulos_renov    = idx_todos[int(n*0.033)  : int(n*0.053)]
idx_texto          = idx_todos[int(n*0.053)  : int(n*0.073)]
idx_outliers       = idx_todos[int(n*0.073)  : int(n*0.083)]
idx_tipos_mezcl    = idx_todos[int(n*0.083)  : int(n*0.098)]
idx_negativos      = idx_todos[int(n*0.098)  : int(n*0.108)]
idx_año_raro       = idx_todos[int(n*0.108)  : int(n*0.113)]
idx_nulos_temp     = idx_todos[int(n*0.113)  : int(n*0.12)]

duplicados = df_sucio.loc[idx_duplicados].copy()
df_sucio = pd.concat([df_sucio, duplicados], ignore_index=True)

df_sucio.loc[idx_nulos_co2, 'co2_emissions'] = np.nan
df_sucio.loc[idx_nulos_renov, 'share_renewables_pct'] = np.nan
df_sucio.loc[idx_nulos_temp, 'temp_anomaly_land'] = np.nan

def ensuciar_texto(valor):
    opciones = [str(valor).upper(), str(valor).lower(), f" {valor} "]
    return np.random.choice(opciones)

df_sucio.loc[idx_texto, 'country'] = df_sucio.loc[idx_texto, 'country'].apply(ensuciar_texto)
df_sucio.loc[idx_outliers, 'temp_anomaly_global'] = np.random.uniform(50, 100, size=len(idx_outliers))
df_sucio.loc[idx_tipos_mezcl, 'primary_energy_twh'] = df_sucio.loc[idx_tipos_mezcl, 'primary_energy_twh'].astype(str) + ' TWh'
df_sucio.loc[idx_negativos, 'share_renewables_pct'] = df_sucio.loc[idx_negativos, 'share_renewables_pct'] * -1
df_sucio.loc[idx_año_raro, 'year'] = 2099

print("=" * 60)
print("1) DIMENSIONES")
print("=" * 60)
print(f"   {df_sucio.shape[0]} filas × {df_sucio.shape[1]} columnas")

print("\n" + "=" * 60)
print("2) TIPOS Y DATOS PRESENTES")
print("=" * 60)
df_sucio.info()

print("\n" + "=" * 60)
print("3) DATOS FALTANTES POR COLUMNA")
print("=" * 60)
print(df_sucio.isna().sum())

print("\n" + "=" * 60)
print("4) FILAS DUPLICADAS")
print("=" * 60)
print(f"   {df_sucio.duplicated().sum()} filas repetidas")

print("\n" + "=" * 60)
print("5) ESTADÍSTICOS DE LAS COLUMNAS NUMÉRICAS")
print("=" * 60)
print(df_sucio.describe())

print("\n" + "=" * 60)
print("6) VALORES ÚNICOS DE COLUMNAS DE TEXTO")
print("=" * 60)
for columna in ["country", "warming_category", "renewable_transition_stage"]:
    print(f"\n--- {columna}: {df_sucio[columna].nunique()} valores distintos ---")
    print(df_sucio[columna].value_counts())
````

````python
#LIMPIAR
df_final = df_sucio.copy()

# --- 1. Duplicados ---
print("\n" + "=" * 60)
print("PASO 1: Duplicados")
print("=" * 60)
print(f"   Duplicados antes: {df_final.duplicated().sum()}")
df_final = df_final.drop_duplicates()
print(f"   Duplicados después: {df_final.duplicated().sum()}")

# --- 2. Año fuera de rango ---
print("\n" + "=" * 60)
print("PASO 2: Año fuera de rango (2004-2024)")
print("=" * 60)
fuera_rango = (df_final['year'] < 2004) | (df_final['year'] > 2024)
print(f"   Filas con año inválido: {fuera_rango.sum()}")
df_final = df_final[~fuera_rango]

# --- 3. Texto inconsistente en 'country' ---
print("\n" + "=" * 60)
print("PASO 3: Texto inconsistente en 'country'")
print("=" * 60)
print(f"   Valores únicos antes: {df_sucio['country'].nunique()}")
df_final['country'] = df_final['country'].str.strip().str.title()
print(f"   Valores únicos después: {df_final['country'].nunique()}")

# --- 4. Tipos mezclados en 'primary_energy_twh' ---
print("\n" + "=" * 60)
print("PASO 4: Tipos mezclados en 'primary_energy_twh'")
print("=" * 60)
print(f"   Tipo antes: {df_sucio['primary_energy_twh'].dtype}")
df_final['primary_energy_twh'] = df_final['primary_energy_twh'].astype(str).str.replace(' TWh', '', regex=False)
df_final['primary_energy_twh'] = pd.to_numeric(df_final['primary_energy_twh'])
print(f"   Tipo después: {df_final['primary_energy_twh'].dtype}")

# --- 5. Negativos inválidos en 'share_renewables_pct' ---
print("\n" + "=" * 60)
print("PASO 5: Negativos inválidos en 'share_renewables_pct'")
print("=" * 60)
print(f"   Valores negativos antes: {(df_final['share_renewables_pct'] < 0).sum()}")
df_final['share_renewables_pct'] = df_final['share_renewables_pct'].abs()
print(f"   Valores negativos después: {(df_final['share_renewables_pct'] < 0).sum()}")

# --- 6. Outliers en 'temp_anomaly_global' ---
print("\n" + "=" * 60)
print("PASO 6: Valores atípicos en 'temp_anomaly_global'")
print("=" * 60)
print("--- ANTES ---")
print(df_final['temp_anomaly_global'].describe())

imposibles = df_final['temp_anomaly_global'] > 3
print(f"\n   Filas con valores imposibles: {imposibles.sum()}")
df_final.loc[imposibles, 'temp_anomaly_global'] = np.nan

print("\n--- DESPUÉS ---")
print(df_final['temp_anomaly_global'].describe())

# --- 7. Imputación de nulos (mediana por país y año) ---
print("\n" + "=" * 60)
print("PASO 7: Imputación de nulos (mediana por país y año)")
print("=" * 60)
columnas_a_imputar = ['co2_emissions', 'share_renewables_pct', 'temp_anomaly_land', 'temp_anomaly_global']
for col in columnas_a_imputar:
    nulos_antes = df_final[col].isna().sum()
    df_final[col] = df_final.groupby(['country', 'year'])[col].transform(lambda x: x.fillna(x.median()))
    nulos_despues = df_final[col].isna().sum()
    print(f"   {col}: {nulos_antes} → {nulos_despues} nulos")

````

## **Hallazgo de patrones, relaciones e insights**
Del análisis exploratorio se identificaron varios patrones relevantes para responder la pregunta de investigación.

**Aceleración reciente del calentamiento global**

Las cuatro variables analizadas (temperatura global, CO₂, calor oceánico y temperatura ártica) muestran una tendencia creciente sostenida durante todo el periodo 2004-2024. Sin embargo, el crecimiento no es constante: entre 2022 y 2024 se observa un salto mucho más pronunciado que en años anteriores, especialmente en la temperatura global (de 0.8°C a 1.6°C) y en la temperatura ártica (de 1.7°C a 3.9°C). Esto sugiere que el calentamiento no solo continúa, sino que se está acelerando en los años más recientes del periodo estudiado.

**Fuerte relación entre temperatura, océano y hielo**

La matriz de correlación confirma el mecanismo físico detrás del problema planteado: la anomalía de temperatura global correlaciona casi perfectamente con la anomalía de temperatura ártica (0.96) y con la anomalía de temperatura de la superficie del mar (0.98), y de forma fuerte con el calor oceánico (0.77) y con el riesgo climático (0.80). Esto respalda la hipótesis inicial del proyecto: el aumento de temperatura global está estrechamente ligado al calentamiento de los océanos y, a través de este, al derretimiento del hielo (aproximado con la anomalía ártica).

**Relación débil entre CO₂/GEI y las variables de temperatura**

Un hallazgo inesperado es que las emisiones de CO₂ y los gases de efecto invernadero totales prácticamente no correlacionan con la temperatura (0.03-0.04) ni con el riesgo climático (0.03) dentro del conjunto de datos. Esto no significa que el CO₂ no cause calentamiento, sino que su efecto opera de forma acumulada a nivel global a lo largo de décadas, mientras que el dataset compara países y años puntuales: un país con emisiones altas en un año determinado no necesariamente presenta mayor temperatura ese mismo año, porque el calentamiento responde a la acumulación histórica de gases y no a la emisión anual de un país específico. Este hallazgo es clave porque anticipa por qué el modelo de regresión (que usa CO₂ y energía para predecir riesgo climático) tiene un desempeño bajo, tal como se explica en la siguiente sección.

**Diferencias entre continentes: calentamiento sincronizado, pero riesgo desigual**

Al comparar la anomalía de temperatura por continente, las seis líneas (África, Asia, Europa, América del Norte, América del Sur y Oceanía) prácticamente se superponen a lo largo de todo el periodo. Esto indica que el aumento de temperatura es un fenómeno global y sincronizado, no algo que ocurra de forma diferenciada por región. Sin embargo, al observar el riesgo climático por continente, sí aparecen diferencias claras: África y Asia presentan medianas más altas y más países-año en niveles de riesgo elevado, mientras que Europa y Oceanía muestran valores más bajos. Esto sugiere que, aunque todos los continentes se calientan al mismo ritmo, el riesgo climático depende también de factores adicionales de vulnerabilidad y exposición, no únicamente de la temperatura.

**Renovables y riesgo climático**

La participación de energías renovables (share_renewables_pct) muestra una correlación positiva débil con el riesgo climático (0.19-0.23), un resultado contraintuitivo. Una posible explicación es que los países que más han avanzado en su transición energética suelen ser economías desarrolladas con un historial largo de emisiones acumuladas, por lo que ya arrastran un riesgo climático alto pese a estar reduciendo sus emisiones actuales.

## **Selección del modelo y justificación de su elección**

Para responder con el análisis de los datos, se desarrollaron dos modelos de predicción: una regresión lineal múltiple y un Random Forest.

Ambos modelos se seleccionaron porque funcionan adecuadamente con las variables numéricas y categóricas presentes en el dataset, sin necesidad de transformaciones complejas. Además, se complementan entre sí: la regresión lineal múltiple ofrece una predicción numérica y el Random Forest permite obtener una predicción categórica.

Respecto a los demás modelos vistos, la regresión binomial y la regresión logística se descartaron dado que ambos requieren una variable de tipo binario, y ninguna de las variables presentan solo dos categorías. Se descartó K-Means, ya que esta técnica busca agrupar datos sin categorías previas, pero el dataset ya cuenta con variables categóricas debidamente definidas.

Para evaluar ambos modelos, el dataset se dividió en dos partes, el 80% lo utiliza como entrenamiento y el 20% para realizar pruebas, agrupando las observaciones por país para evitar que datos del mismo país aparecieran simultáneamente en ambos conjuntos.

1.  **Regresión lineal múltiple**

Busca estudiar si las variables de emisiones y consumo energético permiten predecir el riesgo climático de un país. Para lograrlo se le suministraron 5 variables predictoras: `co2_emissions`, `total_gh`, `primary_energy_twh`, `land_use_change_co2`, `share_renewables_pct`. Con el objetivo de predecir un número de 0-100 que represente el riesgo climático de un país.

El desempeño del modelo se evaluó mediante el coeficiente de determinación (R²) y el error absoluto medio (MAE).

2.  **Random Forest**

Se entrenó un modelo compuesto de 100 árboles de decisión para determinar si a partir del continente y sus cambios de temperatura se puede predecir la categoría de calentamiento de un país. Como variables predictoras se usaron: `continent`, `temp_anomaly_land`, `temp_anomaly_ocean`, `temp_anomaly_arctic` y `climate_risk_score`. El resultado es una variable categórica, con cuatro niveles posibles: Normal, Elevado, Alto y Extremo.

El desempeño se evaluó mediante la exactitud global (accuracy) y un reporte de clasificación detallado por categoría (precisión, sensibilidad y F1-score).

## **Aplicación del modelo**

````python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import r2_score, mean_absolute_error
import pandas as pd

# 1. Variables predictoras (solo emisiones y energía) y objetivo
X = df_final[['co2_emissions', 'total_ghg', 'primary_energy_twh',
              'land_use_change_co2', 'share_renewables_pct']]
y = df_final['climate_risk_score']

# 2. Dividir por país completo (para no repetir el error de fuga de datos)
splitter = GroupShuffleSplit(test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(X, y, groups=df_final['country']))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

# 3. Crear y entrenar el modelo
modelo_multiple = LinearRegression()
modelo_multiple.fit(X_train, y_train)

# 4. Predecir sobre datos que el modelo nunca vio
y_pred = modelo_multiple.predict(X_test)

# 5. Evaluar qué tan bueno es
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
print(f"R²: {r2:.4f}")
print(f"MAE: {mae:.2f}")

# 6. Ver el peso de cada variable
coeficientes = pd.DataFrame({
    'variable': X.columns,
    'coeficiente': modelo_multiple.coef_
}).sort_values('coeficiente', key=abs, ascending=False)
print("\nCoeficientes:")
print(coeficientes)
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,6))
plt.scatter(y_test, y_pred, alpha=0.4, color='steelblue')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2, label='Predicción perfecta')
plt.xlabel('Riesgo climático real')
plt.ylabel('Riesgo climático predicho')
plt.title('Regresión Múltiple: Predicho vs. Real')
plt.legend()
plt.tight_layout()
plt.savefig('regresion_pred_vs_real.png', dpi=150)
plt.show()


plt.figure(figsize=(7,5))
coef_ordenados = coeficientes.sort_values('coeficiente')
colores = ['lightsteelblue' if x < 0 else 'steelblue' for x in coef_ordenados['coeficiente']]
plt.barh(coef_ordenados['variable'], coef_ordenados['coeficiente'], color=colores)
plt.xlabel('Coeficiente')
plt.title('Peso de cada variable en el riesgo climático')
plt.tight_layout()
plt.savefig('regresion_coeficientes.png', dpi=150)
plt.show()
````

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder

# 1. Convertir 'continent' (texto) a número
le = LabelEncoder()
df_final['continent_num'] = le.fit_transform(df_final['continent'])

# 2. Variables predictoras y objetivo
X = df_final[['continent_num', 'temp_anomaly_land', 'temp_anomaly_ocean',
              'temp_anomaly_arctic', 'climate_risk_score']]
y = df_final['warming_category']

# 3. Dividir por país completo
splitter = GroupShuffleSplit(test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(X, y, groups=df_final['country']))
X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

# 4. Crear y entrenar el modelo (100 árboles)
modelo_rf = RandomForestClassifier(n_estimators=100, random_state=42)
modelo_rf.fit(X_train, y_train)

# 5. Predecir y evaluar
y_pred = modelo_rf.predict(X_test)
exactitud = accuracy_score(y_test, y_pred)
print(f"Exactitud: {exactitud:.2%}")
print("\nReporte por categoría:")
print(classification_report(y_test, y_pred))

# 6. Importancia de cada variable
importancias = pd.DataFrame({
    'variable': X.columns,
    'importancia': modelo_rf.feature_importances_
}).sort_values('importancia', ascending=False)
print("\nImportancia de cada variable:")
print(importancias)
````

````python
plt.figure(figsize=(7,5))
imp_ordenadas = importancias.sort_values('importancia')
plt.barh(imp_ordenadas['variable'], imp_ordenadas['importancia'], color='steelblue')
plt.xlabel('Importancia')
plt.title('Importancia de variables — Categoría de calentamiento')
plt.tight_layout()
plt.savefig('rf_importancias.png', dpi=150)
plt.show()


from sklearn.metrics import ConfusionMatrixDisplay

plt.figure(figsize=(6,6))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap='Blues', xticks_rotation=45)
plt.title('Matriz de confusión — Categoría de calentamiento')
plt.tight_layout()
plt.savefig('rf_matriz_confusion.png', dpi=150)
plt.show()
````

 **Comparación entre ambos modelos**

Los resultados contrastantes entre los dos modelos tienen sentido a la luz del análisis exploratorio: la regresión lineal falla porque intenta predecir un índice compuesto (riesgo climático) a partir de variables de emisión que casi no correlacionan con él, mientras que el Random Forest funciona muy bien porque la categoría de calentamiento está directamente determinada por las mismas variables de temperatura que se le entregaron como predictoras. Esto también deja una lección para el proyecto: la selección de variables predictoras es tan importante como la elección del algoritmo, y un buen desempeño no implica automáticamente que el modelo esté "descubriendo" una relación causal profunda.

Adicionalmente, se identificó que el calentamiento se está acelerando: las cuatro variables climáticas analizadas muestran una tendencia sostenida al alza durante todo el periodo, con un salto especialmente pronunciado entre 2022 y 2024. La anomalía de temperatura ártica creció de 1,2°C a 3,4°C, un ritmo notablemente más rápido que el promedio global (0,6°C a 1,6°C), consistente con el fenómeno físico de amplificación polar. Por otro lado, aunque el calentamiento es un fenómeno global sincronizado entre continentes, el riesgo climático no se reparte de forma equitativa: África y Asia presentan los niveles más altos, lo que indica que la vulnerabilidad depende también de factores socioeconómicos y de exposición, no únicamente de la temperatura.

Los modelos desarrollados confirman estos hallazgos del análisis exploratorio. La regresión lineal múltiple, que intentó predecir el riesgo climático a partir de variables de emisiones y energía, obtuvo un desempeño muy bajo (R² = 0,04), reforzando la idea de que el riesgo climático de un país es una consecuencia acumulada a lo largo de muchos años y no algo explicado por sus emisiones actuales. En cambio, el Random Forest, que usó variables de temperatura para predecir la categoría de calentamiento, alcanzó una exactitud del 94%, mostrando que las anomalías térmicas del océano y la tierra sí son buenas predictoras de qué tan extremo es el calentamiento de un país.
